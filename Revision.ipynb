```python
A = np.array(data['Salary'])
B = np.array(data['YearsExperience'])
plt.scatter(A,B)
plt.xlabel("Salary")
plt.ylabel("YearsExperience")
plt.show()
```

```python
slope , intercept , r , p ,std_err = stats.linregress(A,B)

def myfunc(x):
    return slope*x + intercept

mymodel = list(map(myfunc,A))
plt.scatter(A,B)
plt.plot(x,mymodel)

plt.show()
```

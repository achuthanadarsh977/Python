```python
#Dice Rolling
import random

playing = True

while playing:
      choice = input('Enter choice(y/n):')
    
      if choice == 'y':
          dice1 = random.randint(1,6)
          dice2 = random.randint(1,6)
          print([dice1,dice2])

      elif choice == 'n':
          print('Thanks for playing')
          break

      else:
          print('Invalid Choice')
```

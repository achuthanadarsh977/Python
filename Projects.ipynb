```python
#Grade Calculator

john_samuel = {'name': 'John Samuel' , 'assignment' : [50,60,70,80,90] , 'lab' : [90,85,67,76,70] , 'test': [45,56,67,78,90]}

jack_daniels = {'name':'Jack Daniels' , 'assignment' : [60,70,80,80,50] , 'lab' : [86,89,84,76,70] , 'test': [56,67,78,78,90]}

sarah_connor = {'name':'Sarah Connor' , 'assignment' : [90,75,86,73,80] , 'lab' : [79,89,90,78,80] , 'test' : [65,85,84,93,76]}

def get_average(student):
    total = sum(student)
    count = len(student)
    return total/count

def calculate_average(marks):
    letter1 = get_average(marks['assignment'])
    letter2 = get_average(marks['lab'])
    letter3 = get_average(marks['test'])

    return (letter1+letter2+letter3)/3

def calculate_grade(marks):

    if marks >= 91 and marks <= 100:
        return 'A+'

    elif marks >= 81 and marks <= 90:
         return 'A-'

    elif marks >= 71 and marks <= 80:
         return 'B+'

    elif marks >= 61 and marks <= 70:
         return 'B-'

    elif marks >= 51 and marks <= 60:
         return 'C-'

    elif marks >= 40 and marks <= 50:
        return 'D'

    else:
        return 'E'

print('Students')
print('1.John Samuel\n2.Sarah Connor\n3.Jack Daniels')

choice = int(input('Enter number:'))


if choice == 1:
       print('Name:',john_samuel['name'])
       average1 = get_average(john_samuel['assignment'])
       average2 = get_average(john_samuel['lab'])
       average3 = get_average(john_samuel['test'])
       print('\nTotal Average:',calculate_average(john_samuel))
       print('\nFinal Grade:',calculate_grade(calculate_average(john_samuel)))

elif choice == 2:
         print('Name:',sarah_connor['name'])
         print('\nTotal Average:',calculate_average(sarah_connor))
         print('\nFinal Grade:',calculate_grade(calculate_average(sarah_connor)))

elif choice == 3:
         print('Name:',jack_daniels['name'])
         print('\nTotal Average:',calculate_average(jack_daniels))
         print('\nFinal Grade:',calculate_grade(calculate_average(jack_daniels)))

else:
    print('Invalid Choice')

    
         
```

```python
#Dice Rolling 

import random

choice = input('Enter choice(y/n):')

playing = True

while playing:

      if choice == 'y':
          dice1 = random.randint(1,6)
          dice2 = random.randint(1,6)
          print([dice1,dice2])

      elif choice == 'n':
          print('Thanks for playing')
          break

      else:
          print('Invalid Choice')

     
```

```python
#Dice Rolling
import random

playing = True

while playing:
      choice = input('Enter choice(y/n):')
    
      if choice == 'y':
          dice1 = random.randint(1,6)
          dice2 = random.randint(1,6)
          print([dice1,dice2])

      elif choice == 'n':
          print('Thanks for playing')
          break

      else:
          print('Invalid Choice')
```

```python
#NumberToGuess
import random
number_to_guess = random.randint(1,100)

while True:
    try:
        guess = int(input('Enter number to guess:'))
        if number_to_guess > guess:
           print('Too Low')
        elif number_to_guess < guess:
             print('Too High')
        else:
           print('Thank you for entering valid number')
           break

    except ValueError:
           print('Please enter valid number')
```

```python
enter_age = 18

while True:
    try:
      age = int(input('Enter age:'))
      if age >= enter_age:
         print('Eligible to vote')
         break
      else:
         print('Not eligible to vote')
    except ValueError:
        print('Enter valid age')
```

```python
#Currency Converter
currencies = {'USD':1.24 , 'EURO':2.23 , 'GBP':0.19 , 'JPY':35.81 , 'CHF':0.22 , 'CNY':1.73,'AED':0.88}

while True:
    try:
        amount = int(input('Enter amount:'))
        currency = input('Enter currency:')
        if currency in currencies:
            print('Amount converted:',amount*currencies[currency])
        else:
            print('No Currency')
            break
    except ValueError:
         print('Enter valid number')
```

```python
#Rock,Paper,Scissors
import random
choices = ('r','p','s')
dictionary = {'r':'Rock' , 'p':'Paper', 's':'Scissors'}
computer_choice = random.choice(choices)
user_choice = input('Rock , Paper and Scissors:')

if user_choice in choices:

    if user_choice == computer_choice:
        print('Tie')

    elif user_choice == 'r' and computer_choice == 's':
          print('You win')

    elif user_choice == 's' and computer_choice == 'p':
         print('You win')

    elif user_choice == 'p' and computer_choice == 'r':
         print('You win')

    elif user_choice == 'p' and computer_choice == 's':
         print('You lose')

    elif user_choice == 'r' and computer_choice == 'p':
         print('You lose')

    elif user_choice == 's' and computer_choice == 'r':
         print('You lose')

else:
    print('Invalid Choice')
        
```

```python
#Quiz Game
import random

quiz = [ {'question' : 'What is the largest country in the world by land area?' , 'options':['A.China' , 'B.Canada' , 'C.USA' , 'D.Russia'] 
          , 'answer' : 'D'} , {'question':'What is the chemical symbol for gold?' , 'options':['A.Go','B.Gd','C.Au','D.Ag'] , 'answer':'C'},
       
      {'question': 'What does the len() function do in Python?',
       'options': ['A.Deletes an element from a list', 'B.Returns the length of an object',
                   'C.Converts a string to lowercase', 'D.Loops through a list'],
       'answer': 'B'} , 
      {'question': 'What is Pi (π) to two decimal places?',
       'options': ['A.3.12', 'B.3.14', 'C.3.16', 'D.3.41'],
       'answer': 'D'},  
        {'question': 'Who co-founded Apple Inc.?',
       'options': ['A.Bill Gates', 'B.Elon Musk', 'C.Steve Jobs', 'D.Jeff Bezos'],
       'answer': 'C'},

        {'question':'What is the powerhouse of the cell?',
         'options':['A.Nucleus' , 'B.Ribosome' , 'C.Mitochondria' , 'D.Plasma'],
         'answer':'C' 
        },

        {'question':'What does "overfitting" mean?',
         'options':['A.The model is too simple to learn patterns' , 'B.The model performs well on training data but poorly on new data' , 'C. The model takes too long to train' , 'D.The dataset is too large'],
         'answer':'B' 
        },

        {'question':'What is the national language of India?',
         'options':['A.Hindi' , 'B.Tamil' , 'C.Malayalam' , 'D.Telugu'],
         'answer':'A'
        },

        {
            'question':'How many continents are there in the world?',
            'options': ['A.4' , 'B.5' , 'C.6' , 'D.7'],
            'answer':'D'
        },

        {
            'question':'In what year did World War II end?',
            'options':['A.1943' , 'B.1944' , 'C.1945' , 'D.1946'],
            'answer':'C'
        }
        
       ] 

random.shuffle(quiz)

score = 0

for item,i in enumerate(quiz,1):
    print(f'Question {item} : {i['question']}')
    for option in i['options']:
        print(option)

    answer = input('Enter your answer:').upper()
    if answer == i['answer']:
       print('Correct Answer')
       score+=1

    else:
        print('Wrong Answer')

print(f'Game Over , Final Score:{score}/{len(quiz)}')
  
```

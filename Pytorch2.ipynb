```python
 # ── 2. Prepare Features and Target ───────────────────────────
  # Features: distance, time_of_day, is_weekend
X = torch.tensor(df[['distance_miles', 'time_of_day_hours', 'is_weekend']].values,
                   dtype=torch.float32)   # shape → (100, 3)

  # Target: delivery_time_minutes
y = torch.tensor(df['delivery_time_minutes'].values,
                   dtype=torch.float32).unsqueeze(1)  # shape → (100, 1)

print("X shape:", X.shape)   # torch.Size([100, 3])
print("y shape:", y.shape)   # torch.Size([100, 1])

  # ── 3. Normalize Features ─────────────────────────────────────
X = (X - X.mean(dim=0)) / X.std(dim=0)

  # ── 4. Create DataLoader ──────────────────────────────────────
dataset = TensorDataset(X, y)
load    = DataLoader(dataset, batch_size=32, shuffle=True)

  # ── 5. Define Model ───────────────────────────────────────────
model = nn.Sequential(
      nn.Linear(3, 64),    # 3 input features
      nn.ReLU(),
      nn.Linear(64, 32),
      nn.ReLU(),
      nn.Linear(32, 1)     # 1 output: delivery time
)

  # ── 6. Loss and Optimizer ─────────────────────────────────────
loss_fn   = nn.MSELoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)

  # ── 7. Training Loop ──────────────────────────────────────────
for epoch in range(500):
    model.train()
    epoch_loss = 0

    for X_batch, y_batch in load:
        optimizer.zero_grad()
        predictions = model(X_batch)
        loss        = loss_fn(predictions, y_batch)
        loss.backward()
        optimizer.step()
        epoch_loss += loss.item()

    if (epoch + 1) % 50 == 0:
          avg_loss = epoch_loss / len(load)
          print(f"Epoch [{epoch+1}/500]  Loss: {avg_loss:.4f}")

  # ── 8. Test Prediction ────────────────────────────────────────
model.eval()
with torch.no_grad():
      sample = torch.tensor([[5.5, 8.5, 0]], dtype=torch.float32)
      sample = (sample - X.mean(dim=0)) / X.std(dim=0)  # normalize
      pred   = model(sample)
      print(f"\nPredicted delivery time: {pred.item():.2f} minutes")

  
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import accuracy_score
```

```python
data = pd.read_csv(r'C:\Users\SriniAchuthan\OneDrive\Desktop\Machine-Learning-A-Z-Codes-Datasets\Machine Learning A-Z\Part 2 - Regression\Section 8 - Decision Tree Regression\Python\Position_Salaries.csv')
print(data)
```

```python
X = data.iloc[:,1:-1].values
y = data.iloc[:,-1].values
regressor = DecisionTreeRegressor(random_state = 0)
regressor.fit(X,y)
regressor.predict([[6.5]])

X_grid = np.arange(min(X) , max(X) , 0.01)
X_grid = X_grid.reshape((len(X_grid), 1))
plt.scatter(X, y, color = 'red')
plt.plot(X_grid, regressor.predict(X_grid), color = 'blue')
plt.title('Truth or Bluff (Decision Tree Regression)')
plt.xlabel('Position level')
plt.ylabel('Salary')
plt.show()
```

```python
regressor = RandomForestRegressor(n_estimators = 10 , random_state = 0)
regressor.fit(X,y)
regressor.predict([[6.5]])

X_grid = np.arange(min(X) , max(y) , 0.01)
X_grid = X_grid.reshape((len(X_grid),1))
plt.scatter(X, y, color = 'red')
plt.plot(X_grid, regressor.predict(X_grid), color = 'blue')
plt.title('Truth or Bluff (Random Forest Regression)')
plt.xlabel('Position level')
plt.ylabel('Salary')
plt.show()
```

```python
from sklearn.cluster import KMeans
kmeans = KMeans(n_clusters = 2 , random_state = 42 , n_init = 10)
np1 = np.array([
      [25, 80],
      [30, 85],
      [22, 90],
      [45, 20],
      [50, 15],
      [48, 25],
      [35, 50],
      [32, 55],
      [60, 10],
      [28, 75]
  ])
print(np1)
```

```python
kmeans = KMeans(n_clusters=2, random_state=42, n_init=10)
kmeans.fit(np1)

labels    = kmeans.labels_
centroids = kmeans.cluster_centers_
inertia   = kmeans.inertia_

print("Shape:    ", np1.shape)
print("Labels:   ", labels)
print("Centroids:", centroids)
print("Inertia:  ", inertia)

  # Plot — no manual color indexing, avoids the error
plt.scatter(np1[:, 0], np1[:, 1], c=labels, cmap='bwr', s=100)

for i, point in enumerate(np1):
      plt.annotate("ABCDEFGHIJ"[i], (point[0], point[1]),
                   textcoords="offset points", xytext=(5, 5))

plt.scatter(centroids[:, 0], centroids[:, 1],
              marker='X', s=200, color='black', label='Centroids')
plt.xlabel("Age")
plt.ylabel("Spending Score")
plt.title("K-Means Clustering")
plt.legend()
plt.show()

```

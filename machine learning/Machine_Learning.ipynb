```python
X = (X - X.mean()) / X.std()
w = 0.0
b = 0.0
alpha = 0.01
epoch = 1000
m = len(X)

for epoch in range(epoch):

     y_pred = w * X + b

      # Step 2: Compute gradients
     dw = (1/m) * np.sum((y_pred - y) * X)
     db = (1/m) * np.sum(y_pred - y)

      # Step 3: Update parameters
     w = w - alpha * dw
     b = b - alpha * db

      # Step 4: Track cost (optional)
     if epoch % 100 == 0:
          cost = (1/(2*m)) * np.sum((y_pred - y) ** 2)
          print(f"Epoch {epoch} | Cost: {cost:.2f} | w: {w:.2f} | b: {b:.2f}")

print(f"\nFinal → w: {w:.2f}, b: {b:.2f}")
```

```python
data3 = pd.read_csv(r'C:\Users\SriniAchuthan\OneDrive\Desktop\Machine-Learning-A-Z-Codes-Datasets\Machine Learning A-Z\Part 2 - Regression\Section 5 - Multiple Linear Regression\Python\50_Startups.csv')
print(data3)
```

```python
X = data3[['R&D Spend']].values
y = data3['Administration'].values

scaler_X = StandardScaler()
scaler_y = StandardScaler()

X_scaled = scaler_X.fit_transform(X)
y_scaled = scaler_y.fit_transform(y.reshape(-1,1)).ravel()


svr = SVR(kernel='rbf', C=100, epsilon=0.1, gamma=0.1)
svr.fit(X_scaled, y_scaled)



#   # ── Step 3: Predict on training data ──────────────────
y_pred_scaled = svr.predict(X_scaled)
y_pred = scaler_y.inverse_transform(y_pred_scaled.reshape(-1, 1)).ravel()

  # ── Step 4: Plot ───────────────────────────────────────
X_grid = np.linspace(1, 10, 300).reshape(-1, 1)
X_grid_scaled = scaler_X.transform(X_grid)
y_grid_pred = scaler_y.inverse_transform(
svr.predict(X_grid_scaled).reshape(-1, 1)
  ).ravel()

plt.figure(figsize=(8, 5))
plt.scatter(X, y, color='green', zorder=5, label='Actual Salary')
plt.plot(X_grid, y_grid_pred, color='red', linewidth=2, label='SVR Fit')
plt.xlabel('R&D Spend')
plt.ylabel('Administration')
plt.title('SVR — R&D Spend vs Administration')
plt.xticks(range(1, 11))
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

  # ── Step 5: Predict a new level ───────────────────────
level = np.array([[10000]])
level_scaled = scaler_X.transform(level)
pred_scaled = svr.predict(level_scaled)
pred_salary = scaler_y.inverse_transform(pred_scaled.reshape(-1, 1))
print(f"\nPredicted Amount for Level 10000: ${pred_salary[0][0]:,.0f}")
```

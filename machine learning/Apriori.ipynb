```python
dataset = pd.read_csv('supermarket.csv', header=None)
transactions = dataset.fillna('').values.tolist()
transactions = [[str(item).strip() for item in row if str(item).strip() != ''] for row in transactions]

total_transactions = len(transactions)

   # Build vertical TID list
tid_list = {}
for tid, transaction in enumerate(transactions):
    for item in transaction:
        if item not in tid_list:
               tid_list[item] = set()
        tid_list[item].add(tid)

print('--- Vertical TID List ---')
for item, tids in sorted(tid_list.items()):
    print(f'{item:10} -> transactions: {sorted(tids)}  support: {len(tids)/total_transactions:.2f}')
print()

   # ECLAT function
def eclat(prefix, items, min_support):
    for i, (item_i, tidset_i) in enumerate(items):
        support = len(tidset_i) / total_transactions
        if support >= min_support:
            yield prefix + [item_i], support

            suffix = []
            for item_j, tidset_j in items[i+1:]:
                new_tidset = tidset_i & tidset_j
                if len(new_tidset) / total_transactions >= min_support:
                    suffix.append((item_j, new_tidset))

            yield from eclat(prefix + [item_i], suffix, min_support)

   # Run ECLAT
min_support = 0.2
items = sorted(tid_list.items())
results = list(eclat([], items, min_support))

   # Display results
print('--- Frequent Itemsets (min_support=0.2) ---')
df = pd.DataFrame(results, columns=['Itemset', 'Support'])
df['Support'] = df['Support'].round(2)
df = df.sort_values('Support', ascending=False).reset_index(drop=True)
print(df.to_string())
```

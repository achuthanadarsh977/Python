```python
def factorial(n):
    if n < 0:
        return 'Invalid Number'

    elif n == 1 or n == 0:
         return 1

    else:
        return n * factorial(n-1)


print(factorial(5))
    
```

```python
def armstrong(num):
    n = num
    power = len(str(num))
    total = 0

    while n > 0:
          digit = n % 10
          total += digit**power
          n //= 10

    if total == num:
        print('Armstrong number')

    else:
        print('Not Armstrong number')


armstrong(153)
        
    
```

```python
class Calculator:
    def __init__(self,a,b):
        self.a = a
        self.b = b

    def add(self):
        return self.a+self.b

    def product(self):
        return self.a*self.b

    def difference(self):
        return self.a-self.b

    def modulus(self):
        return self.a%self.b

    def divide(self):
        return self.a/self.b

c = Calculator(12344556,1333)
print('Sum of two numbers:',c.add())
print('Difference of two numbers:',c.difference())
print('Product of two numbers:',c.product())
print('Modulus of two numbers:',c.modulus())
print('Quotient of two numbers:',c.divide())
```

```python
#Encapsulation
class BankAccount:

      def __init__(self,balance):
          self.__balance = balance # private 

      def withdraw(self,amount):
          self.amount = amount
          self.__balance-=self.amount # private

      def deposit(self,amount):
          self.amount = amount
          self.__balance+=self.amount

      def get_balance(self):
          return self.__balance

bank = BankAccount(23000)
bank.withdraw(3000)
bank.deposit(1000)
bank.get_balance()
```

```python
class Stack:

    def __init__(self,list):
        self.list = []

    def push(self,adder):

        self.adder = adder
        self.list.append(self.adder)

    def remove(self,number):
        self.number = number
        self.list.remove(self.number)

    def isEmpty(self):
        return len(self.list) == 0

    def pop(self):
        if(self.isEmpty() == True):
            return 'Stack is empty'
        self.list.pop()

    def size(self):
        return len(self.list)


    def peek(self):
        if(self.isEmpty()):
            return 'Stack is empty'
        return self.list[-1]
        

    def display(self):
        return self.list


s = Stack([])
s.push(1)
s.push(23)
s.push(34)
s.push(45)
s.push(64)
s.pop()
print(s.size())
print(s.peek())

s.display()
```

```python
class Queue:

      def __init__(self,queue):
          self.queue = []

      def enqueue(self,number):
          self.number = number
          self.queue.append(self.number)

      def isEmpty(self):
          return len(self.queue) == 0

      def size(self):
          return len(self.queue)

      def peek(self):
          if self.isEmpty() == True:
              return 'Stack is empty'
          return self.queue[0]

      def dequeue(self):
           if self.isEmpty() == True:
              return 'Stack is empty'
           return self.queue.pop(0)
          

      def size(self):
          return len(self.queue)

      def display(self):
          return self.queue

q = Queue([])
q.enqueue(23)
q.enqueue(34)
q.enqueue(43)
q.enqueue(25)
q.enqueue(46)
print(q.display())
q.peek()

      
```

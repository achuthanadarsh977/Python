```python
import ollama
from typing import List, Dict, Any


def generate_with_single_input(prompt: str,
                               role: str = 'user',
                               top_p: float = None,
                               temperature: float = None,
                               max_tokens: int = 500,
                               model: str = "llama3.2:latest",
                               **kwargs):
    options = {}
    if temperature is not None:
        options["temperature"] = temperature
    if top_p is not None:
        options["top_p"] = top_p
    if max_tokens is not None:
        options["num_predict"] = max_tokens

    response = ollama.chat(
        model=model,
        messages=[{"role": role, "content": prompt}],
        options=options
    )
    return {"role": "assistant", "content": response['message']['content']}


def generate_with_multiple_input(messages: List[Dict],
                                 top_p: float = None,
                                 temperature: float = None,
                                 max_tokens: int = 500,
                                 model: str = "llama3.2:latest",
                                 **kwargs):
    options = {}
    if temperature is not None:
        options["temperature"] = temperature
    if top_p is not None:
        options["top_p"] = top_p
    if max_tokens is not None:
        options["num_predict"] = max_tokens

    response = ollama.chat(
        model=model,
        messages=messages,
        options=options
    )
    return {"role": "assistant", "content": response['message']['content']}


def call_llm_with_context(prompt: str, context: list, role: str = 'user', **kwargs):
    context.append({'role': role, 'content': prompt})
    response = generate_with_multiple_input(context, **kwargs)
    context.append(response)
    return response
```
